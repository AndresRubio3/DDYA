## 1) Descripción del problema **[1 pts]**

Se pide calcular el crecimiento de subscriptores en una plataforma de streaming, se determinó que el número de nuevos subscriptores en el mes n sigue el siguiente comportamiento:

• En el mes 0 hay 0 nuevos subscriptores.

• En el mes 1 hay 1 nuevo subscriptor.

• A partir del mes 2, el número de nuevos subscriptores debe ser sub(n-1)+sub(n-2)

Usando memoización se debe calcular cuántos nuevos subscriptores habrán en el mes n.

## 2) Requerimientos **[2 pts]**

- Casos base: el programa debe retornar 0 si el mes es 0, y 1 si el mes es 1.  
- Memoización: el algoritmo debe implementar memoización para optimizar el tiempo.  
- Subscriptores: el programa calculará el número de subscriptores en el mes n.
### Requerimientos No Funcionales

## 6) Código funcional **[15 pts]**

```python
def entero_positivo(texto):
    if len(texto) == 0:
        return False
 
    for caracter in texto:
        if caracter not in "0123456789":
            return False
 
    return True
 
 
def subscriptores(n, memo=None):
    if memo is None:
        memo = {}
 
    if n == 0:
        return 0
    if n == 1:
        return 1

    if n in memo:
        return memo[n]
 
    memo[n] = subscriptores(n - 1, memo) + subscriptores(n - 2, memo)
    return memo[n]
 
 
def main():
    print("  Nuevos subscriptores")
 
    dato = input("Ingrese el mes (debe ser un número entero mayor o igual a 0): ").strip()
 
    if not entero_positivo(dato):
        print("Error: debe ser un numero entero mayor o igual a 0.")
        return
 
    n = int(dato)
 
    memo = {}
    resultado = subscriptores(n, memo)
 
    print(f"\nEn el mes {n} hay {resultado:,} subscriptores.")
 
main()
```
